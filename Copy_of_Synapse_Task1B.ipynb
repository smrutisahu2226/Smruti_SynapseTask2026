**scaler.transform() will return an array. We need to convert this into a dataframe. Do this and add the column names to the dataframe. Call this new dataframe as "df_feat". Call head() on this df**<br>
Note: The final dataframe will have the initial columns except the "TARGET CLASS".

```python
from sklearn.preprocessing import StandardScaler

# Ensure scaler is instantiated and fitted to our features
scaler = StandardScaler()
X = df.drop('TARGET CLASS', axis=1)
X_scaled = scaler.fit_transform(X)

# Convert the scaled array back to a DataFrame with column names
df_feat = pd.DataFrame(X_scaled, columns=df.columns[:-1])

# Call head() on the new dataframe
df_feat.head()
```

## Train/Test split

### Set X to be equal to df_feat and set y accordingly. As you know, X contains our training features and y contains our target.
Hint: y can be taken directly from the initial dataframe "df"

```python
# Set X to be our standardized features and y to be the target class
X = df_feat
y = df['TARGET CLASS']
```

**Import train_test_split function from scikit-learn**<br>
**Create X and y train and test splits in one command using a test size of 0.3 and a random seed**<br>
They should be called X_train, X_test, y_train, y_test

```python
from sklearn.model_selection import train_test_split

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=101
)
```

### Model fit and training

Before moving ahead with the cells below, learn how the K Nearest Neighbour Algorithm actually works.

Theory and implementation : https://youtu.be/wTF6vzS9fy4 <br>
Theory: https://www.analyticsvidhya.com/blog/2018/08/k-nearest-neighbor-introduction-regression-python/

**import KNeighborsClassifier from sklearn and initialize it with neighbours = 1 (more on this later) . Fit this on X_train and y_train**

```python
from sklearn.neighbors import KNeighborsClassifier

# Instantiate the KNN model with n_neighbors=1
knn = KNeighborsClassifier(n_neighbors=1)

# Fit the model to our training data
knn.fit(X_train, y_train)
```

**Using this fitted model, predict on X_test. Store these predictions in variable called pred.**

```python
# Use the fitted model to make predictions on the test set
pred = knn.predict(X_test)

# Display the first few predictions
print(pred[:10])
```

**Let us check how correct these predictions are.<br>
Print a classification report of y_test and pred**<br>
Hint: sklearn classification_report

```python
from sklearn.metrics import classification_report

# Print the classification report
print(classification_report(y_test, pred))
```

#### Learn all about classification metrics - https://medium.com/analytics-vidhya/evaluation-metrics-for-classification-models-e2f0d8009d69

**Print the accuracy using numpy and round it to 3 decimal places.**

```python
import numpy as np

# Calculate accuracy using numpy
accuracy = np.mean(pred == y_test)

# Print the accuracy rounded to 3 decimal places
print(f"Accuracy: {round(accuracy, 3)}")
```

##### Hey, not bad! But can we further improve this? Remember our arbitrary choice for n_neighbours? Can we improve that?

### Choosing optimal 'k'

**Above, we chose n_neighbours to be equal to 1. Choosing a small value of K leads to unstable decision boundaries. <br>
We need to select n_neighbours by calculating the accuracy for every value of n from 1 to 60 and then choosing the one which gives the highest accuracy.**

```python
accuracies = []

# Calculate the test accuracy for every value of K from 1 to 60
for i in range(1, 61):
    knn_model = KNeighborsClassifier(n_neighbors=i)
    knn_model.fit(X_train, y_train)
    pred_i = knn_model.predict(X_test)
    accuracies.append(np.mean(pred_i == y_test))

# Print the first few calculated accuracies
print("Accuracies calculated successfully!")
```

##### **Plot a graph of K value vs Accuracy**

```python
import matplotlib.pyplot as plt

# Create a plot for K values vs Accuracies
plt.figure(figsize=(10, 6))
plt.plot(range(1, 61), accuracies, color='blue', linestyle='dashed', marker='o',
         markerfacecolor='red', markersize=8)

plt.title('Accuracy vs. K Value')
plt.xlabel('K')
plt.ylabel('Test Accuracy')
plt.grid(True)
plt.show()
```

##### **Choose the best value of n_neighbours and give a reason why and also print the accuracy**

```python
# Programmatically find the best K value and its corresponding accuracy
best_k = np.argmax(accuracies) + 1
max_accuracy = max(accuracies)

print(f"Best K value: {best_k}")
print(f"Highest Test Accuracy: {round(max_accuracy, 3)}")
print(f"\nReason: K = {best_k} achieves the absolute peak accuracy on our test set ({round(max_accuracy*100, 1)}%). Choosing this larger K value also helps create a smoother, more stable decision boundary compared to a small value like K = 1.")
```
